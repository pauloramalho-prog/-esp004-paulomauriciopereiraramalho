### Capítulo 1 — Organização do Ambiente de Trabalho

Este notebook verifica que **todo o ambiente** está corretamente instalado:
- Ubuntu 24.04 LTS
- ROS 2 Jazzy Jalisco
- Jupyter com kernel ROS 2
- Python 3.12 e bibliotecas
- Git/GitHub configurados

> **Como usar:** Execute célula por célula (Shift+Enter). Cada célula tem uma cor de saída esperada: 🟢 OK, 🔴 Problema, 🟡 Atenção.

```python
# ═══════════════════════════════════════════════════
# CÉLULA 1 — Verificar Ubuntu 24.04
# ═══════════════════════════════════════════════════
import subprocess, platform

print('='*55)
print('  VERIFICAÇÃO DO SISTEMA OPERACIONAL')
print('='*55)

# Sistema
result = subprocess.run(['lsb_release', '-a'], capture_output=True, text=True)
print(result.stdout)

# Python
print(f'Python: {platform.python_version()}')
py_ver = tuple(int(x) for x in platform.python_version().split('.'))

# Avaliação
if 'Ubuntu 24.04' in result.stdout:
    print('\n🟢 Ubuntu 24.04 LTS detectado — CORRETO!')
elif 'Ubuntu' in result.stdout:
    print('\n🟡 Ubuntu detectado, mas não é 24.04. Verifique com o professor.')
else:
    print('\n🔴 Ubuntu 24.04 NÃO encontrado. Verifique a instalação.')

if py_ver >= (3, 10):
    print(f'🟢 Python {platform.python_version()} — OK (≥3.10 necessário)')
else:
    print(f'🔴 Python {platform.python_version()} — Versão muito antiga!')
```

```python
# ═══════════════════════════════════════════════════
# CÉLULA 2 — Verificar ROS 2 Jazzy
# ═══════════════════════════════════════════════════
import subprocess, os

print('='*55)
print('  VERIFICAÇÃO DO ROS 2 JAZZY')
print('='*55)

# Versão do ros2
r = subprocess.run(['ros2', '--version'], capture_output=True, text=True)
if r.returncode == 0:
    print(f'ros2 version: {r.stdout.strip()}')
    if 'jazzy' in r.stdout.lower():
        print('🟢 ROS 2 Jazzy — CORRETO!')
    else:
        print('🟡 ROS 2 encontrado, mas não é Jazzy:', r.stdout.strip())
else:
    print('🔴 ros2 não encontrado! Execute no terminal:')
    print('   source /opt/ros/jazzy/setup.bash')

# Variáveis de ambiente
ros_dist = os.environ.get('ROS_DISTRO', 'NÃO DEFINIDA')
ros_domain = os.environ.get('ROS_DOMAIN_ID', 'NÃO DEFINIDA')
print(f'\nROS_DISTRO    = {ros_dist}')
print(f'ROS_DOMAIN_ID = {ros_domain}')

if ros_dist == 'jazzy':
    print('🟢 Variáveis de ambiente ROS 2 configuradas corretamente')
else:
    print('🔴 ROS_DISTRO não está configurada! Adicione ao ~/.bashrc:')
    print('   source /opt/ros/jazzy/setup.bash')
    print('   export ROS_DOMAIN_ID=0')
```

```python
# ═══════════════════════════════════════════════════
# CÉLULA 3 — Verificar rclpy e bibliotecas
# ═══════════════════════════════════════════════════
print('='*55)
print('  VERIFICAÇÃO DAS BIBLIOTECAS PYTHON')
print('='*55)

libs = {
    'rclpy':         'ROS 2 Python client',
    'numpy':         'Computação numérica',
    'matplotlib':    'Visualização/gráficos',
    'pandas':        'Análise de dados',
    'torch':         'PyTorch (IA/ML)',
    'rosbags':       'Leitura de bags ROS 2',
    'ipywidgets':    'Widgets interativos Jupyter',
}

all_ok = True
for lib, desc in libs.items():
    try:
        mod = __import__(lib)
        ver = getattr(mod, '__version__', '?')
        print(f'🟢 {lib:<15} {ver:<12} — {desc}')
    except ImportError:
        print(f'🔴 {lib:<15} NÃO INSTALADA  — {desc}')
        print(f'   Instale com: pip3 install {lib}')
        all_ok = False

print()
if all_ok:
    print('🟢 Todas as bibliotecas estão instaladas!')
else:
    print('🔴 Instale as bibliotecas ausentes e reinicie o kernel.')
```

```python
# ═══════════════════════════════════════════════════
# CÉLULA 4 — Verificar Git e GitHub
# ═══════════════════════════════════════════════════
import subprocess

print('='*55)
print('  VERIFICAÇÃO DO GIT E GITHUB')
print('='*55)

# Versão do git
r = subprocess.run(['git', '--version'], capture_output=True, text=True)
print(f'Git: {r.stdout.strip()}')

# Identidade configurada
name  = subprocess.run(['git', 'config', '--global', 'user.name'],
                       capture_output=True, text=True).stdout.strip()
email = subprocess.run(['git', 'config', '--global', 'user.email'],
                       capture_output=True, text=True).stdout.strip()

print(f'\nIdentidade Git:')
print(f'  Nome:  {name or "NÃO CONFIGURADO"}')
print(f'  Email: {email or "NÃO CONFIGURADO"}')

if name and email:
    print('🟢 Identidade Git configurada')
else:
    print('🔴 Configure com:')
    print('   git config --global user.name  "Seu Nome"')
    print('   git config --global user.email "email@ifba.edu.br"')

# Chave SSH
import os
ssh_key = os.path.expanduser('~/.ssh/id_ed25519_github.pub')
if os.path.exists(ssh_key):
    with open(ssh_key) as f:
        key_preview = f.read().strip()[:40] + '...'
    print(f'\n🟢 Chave SSH encontrada: {key_preview}')
    # Testar conexão
    r2 = subprocess.run(['ssh', '-T', 'git@github.com', '-o',
                         'StrictHostKeyChecking=no', '-o', 'ConnectTimeout=5'],
                        capture_output=True, text=True)
    if 'successfully authenticated' in r2.stderr:
        print('🟢 Autenticação GitHub OK!')
    else:
        print('🟡 Chave SSH existe, mas GitHub ainda não autenticou.')
        print('   Adicione a chave em: https://github.com/settings/keys')
else:
    print('\n🔴 Chave SSH não encontrada. Execute no terminal:')
    print('   ssh-keygen -t ed25519 -C "email@ifba.edu.br" -f ~/.ssh/id_ed25519_github')
```

```python
# ═══════════════════════════════════════════════════
# CÉLULA 5 — Teste rclpy: publicar mensagem
# ═══════════════════════════════════════════════════
import rclpy
from std_msgs.msg import String

print('='*55)
print('  TESTE RCLPY — PUBLICAR MENSAGEM')
print('='*55)

try:
    if not rclpy.ok():
        rclpy.init()

    node = rclpy.create_node('teste_jupyter')
    pub  = node.create_publisher(String, '/esp004/teste', 10)

    msg = String()
    msg.data = 'Ambiente ESP004 funcionando! 🟢'
    pub.publish(msg)

    print(f'🟢 Nó criado: /teste_jupyter')
    print(f'🟢 Mensagem publicada em /esp004/teste')
    print(f'   Conteúdo: "{msg.data}"')
    print()
    print('✅ Verifique em outro terminal com:')
    print('   ros2 topic echo /esp004/teste')

    node.destroy_node()
    rclpy.shutdown()

except Exception as e:
    print(f'🔴 Erro: {e}')
    print('Certifique-se de usar o kernel "Python 3 (ROS2 Jazzy)"')
```

```python
# ═══════════════════════════════════════════════════
# CÉLULA 6 — Relatório final de saúde do ambiente
# ═══════════════════════════════════════════════════
import subprocess, os, platform

print('='*55)
print('  RELATÓRIO DE SAÚDE DO AMBIENTE ESP004')
print('='*55)

checks = {}

# Ubuntu
r = subprocess.run(['lsb_release', '-r'], capture_output=True, text=True)
checks['Ubuntu 24.04'] = '24.04' in r.stdout

# ROS 2
checks['ROS 2 Jazzy'] = os.environ.get('ROS_DISTRO') == 'jazzy'

# rclpy
try:
    import rclpy
    checks['rclpy'] = True
except:
    checks['rclpy'] = False

# torch
try:
    import torch
    checks['PyTorch'] = True
except:
    checks['PyTorch'] = False

# Git
name = subprocess.run(['git','config','--global','user.name'],
                      capture_output=True, text=True).stdout.strip()
checks['Git identidade'] = bool(name)

# Exibir
ok_count = sum(checks.values())
total    = len(checks)

for item, ok in checks.items():
    status = '🟢 OK' if ok else '🔴 FALHA'
    print(f'  {status}  {item}')

print()
print(f'Resultado: {ok_count}/{total} verificações passaram')
if ok_count == total:
    print('✅ AMBIENTE COMPLETAMENTE CONFIGURADO!')
    print('   Pode prosseguir para o NB-02.')
else:
    print('⚠️  Resolva os itens com FALHA antes de continuar.')
```
